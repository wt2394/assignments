## Part 3: Functions

```python
def kelvin_to_celsius(K):
    return K - 273.15

def celsius_to_kelvin(C):
    return C + 273.15
```

```python
def to_fahrenheit(temp, unit='C'):
    if unit == 'C':
        return temp * 9/5 + 32
    else: 
        c = kelvin_to_celsius(temp)
        return c * 9/5 + 32 
```

```python
print(to_fahrenheit(0)) 
print(to_fahrenheit(25)) 
```

```python
def from_fahrenheit(temp, unit='C'):
    c = (temp - 32) * 5/9 
    if unit == 'C':
        return c
    elif unit == 'K':
        return c + 273.15
```

```python
def height_to_meters(feet, inches):
    x_inches = feet * 12 + inches
    meters = x_inches * 0.0254
    return meters
```

```python
print(height_to_meters(10, 0))
```

```python
def meters_to_height(meters):
    total_inches = meters / 0.0254
    feet = int(total_inches // 12)
    inches = total_inches % 12
    return feet, inches
```

```python
print(height_to_meters(12,0))
print(meters_to_height(3.6576))
```

```python
print(height_to_meters(3,1))
print(meters_to_height(0.9398))
```

```python
print(height_to_meters(0,5))
print(meters_to_height(0.127))
```
